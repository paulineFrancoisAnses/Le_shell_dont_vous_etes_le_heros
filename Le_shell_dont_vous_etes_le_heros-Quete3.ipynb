## Les conseils de Polymalin 

### Exclure des lignes en début et fin de fichier : _head_ et _tail_ &#x1F451; &#x1F451;

Vous savez parfaitement afficher les lignes mais comment, à l'inverse, ne pas les afficher ? 

Si on traduit le langage humain en langage informatique, "ne pas afficher la dernière ligne" c'est comme "afficher le début du document jusqu'à l'avant-dernière ligne. 

Si mon fichier fait 50 lignes, je veux faire un `head -n 50-1 mon_fichier.txt`. Cependant, pas simple de connaitre le nombre de ligne exact pour chaque fichier et à chaque instant. Alors on peut simplement écrire `-1`.

```bash
# Afficher mon fichier, à l'exception de sa dernière ligne
head -n -1 monfichier.txt
```

Même logique pour afficher un fichier à l'exception de sa première ligne. On part de zéro, on affiche les lignes à partir de la deuxième, équivalent à `tail -n 0+2`. Là encore, on ne peut pas l'écrire comme ça et on ne concerve que le `+2`

```bash
# Afficher mon fichier, à l'exception de sa première ligne
tail -n +2 monfichier.txt
```

## Les conseils de Polymalin 

### Réduire une liste en enlevant la redondance : _uniq_ &#x1F451;

Il arrive qu'une longue liste d'éléments vous soit fournie et que seul son contenu, sans redondance, vous intéresse. Dans ce cas, la commande qu'il vous faut est `uniq`.

Prenons par exemple la liste suivante :
- chat
- chat
- Cheval
- chien
- chien
- Chien

Si vous voulez la liste des animaux présents dans cette liste, vous pouvez faire un `uniq` qui vous fournira le résultat suivant :
- chat
- Cheval
- chien
- Chien

Si *chien* apparaît deux fois, c'est parce que cette commande est sensible à la casse.

Par déduction, quel devrait être la sortie de la commande sur le fichier suivant ?
- chat
- chat
- chien
- chien
- cheval
- chien

```bash
uniq mon_fichier.txt
# chat
# chien
# cheval
# chien
```

`uniq` ne pourra retirer les doublons que si ceux-ci se suivent dans la liste. Sinon, ils seront considérés comme deux élèments distincts. C'est pourquoi il est recommandé d'utiliser la commande `uniq` en tandem avec la commande `sort`.

### Trier une liste par item : _sort_ &#x1F451;

La commande `sort`, comme son nom l'indique, permet de trier une liste ou un tableau. 

Utilisée sans argument, elle trie par défaut selon la première colonne et par ordre alphabétique croissant.

Attention ! En ordre alphabétique, les chiffres se trient comme suit : 1, 10, 100, 2, 20, ...

Les arguments les plus utilisés sont :
- `-r` : Tri en ordre inverse (r : _reverse_)
- `-n` : Tri par ordre alphanumérique (les chiffres sont correctement pris en compte, le 9 vient donc avant le 10)
- `-k x,x` : Tri sur la colonne x. 

```bash
# Tri standard
sort mon_fichier.txt
# Tri selon la colonne 4 en ordre alphanumérique, puis selon la colonne 2 en ordre inverse
sort -k4,4n -k 2,2r
```

### Afficher du texte à l'écran : _echo_ et variables &#x1F451;

La majorité des bioinformagiciens vous dira que `echo` est la commande de base. Mais à quoi bon l'apprendre si vous n'en voyez pas de suite l'intérêt ?

Vous êtes suffisamment avancé pour avoir un réel besoin de cette commande qui ne fait que... imprimer à l'écran ce qui la suit.

```bash
echo "Bonjour"
# Le terminal écrit "Bonjour"
echo ""
# Le terminal fait un simple retour à la ligne
```

Mais `echo` a bien plus à offrir ! Pour cela, vous allez devoir découvrir les variables.

Une variable est une chaîne de caractères dans laquelle vous stockez une valeur. N'importe quelle valeur.

Il vous suffit de la définir puis de l'utiliser.

Un exemple vaut mille mots :

```bash
# Définition de la variable "elfe" valant "elfe_046"
elfe="elfe_046"
# Impression de la valeur de la variable dans le terminal
echo ${elfe}
# Sans surprise, le résultat à l'écran sera "elfe_046"
```

La syntaxe d'appel de variable `${}` est un peu lourde mais elle vous sauvera la vie (ok peut-être pas la vie mais à minima des demis-journées de prise de tête).

### Tester une condition : _if_ &#x1F451; &#x1F451;

Il va falloir commencer à vous accrocher. Cette notion n'est pas simple mais je vais vous la décomposer et y aller pas à pas !

#### Principe du `if` : 

- En français cela donne "_SI_ ma condition est vraie _ALORS_ je réalise une action _SINON_ je réalise une autre action".

- En anglais maintenant : _IF_ ma condition est vraie _THEN_ je réalise une action _ELSE_ je réalise une autre action.

Exemple : __*If*__ il fait beau, __*then*__ je sors faire un tour dans la forêt, __*else*__ je regarde la boîte magique assis sur mon tabouret à 4 pieds.

#### En code : 

`if [` ma condition est vraie `]; then` je réalise une action`; else` je réalise une autre action`; fi`

Dans l'exemple précédent : 

`if [` temps=beau `]; then` sortir \${foret} `; else` rester \${maison}`;` regarder ${boite_magique} `;` asseoir \${tabouret_a_4_pieds} `; fi` 

#### Pour aller plus loin :

Dans l'éventualité où vous souhaiteriez tester une condition, puis une autre, vous pouvez utiliser `elif` :

`if` `[` ma condition est vraie `]; then` je réalise une action`; elif [` nouvelle condition de test `]; then` je réalise une autre action`; else` je réalise encore une autre action `; fi`


Nouvel exemple traduit en franglais : __*If*__ il fait beau, __*then*__ je sors faire un tour dans la forêt, __*elif (else if)*__ il fait nuageux, __*then*__ je sors en prenant un gilet, __*else*__ je regarde la boîte magique assis sur mon tabouret à 4 pieds.

Reste maintenant à remplir les blancs !

### Tester une condition : _if_ &#x1F451; &#x1F451; &#x1F451;

Maintenant que vous avez vu la structure d'un `if`, voyons quelles conditions vous pouvez tester.

#### Un fichier existe-t-il ?

Rien de plus simple, il faudra mettre l'option `-f` à l'intérieur des crochets :

```bash
# Tester si mon_fichier.txt existe et si oui, afficher "Mon fichier existe" à l'écran
if [ -f mon_fichier.txt ]; then echo "Mon fichier existe"; fi
# Tester si mon_fichier.txt existe et si oui, j'écris dedans et à la suite et s'il n'existe pas j'écris dans un fichier ayant ce même nom.
if [ -f mon_fichier.txt ]; then echo "dis moi pourquoi j'existerais ?" >> mon_fichier.txt; else echo "Et si tu n'existais pas..." > mon_fichier.txt; fi
```

#### Un dossier existe-t-il ?

Même chose que pour un fichier mais avec l'option `-d`.

```bash
# Tester si mon dossier existe et si oui, lister son contenu
if [ -d mon_dossier ]; then ls mon_dossier; fi
```

#### Une variable est-elle vide ?

Cette fois c'est l'option `-z` qui vous permettra de tester si votre variable est vide.

```bash
# Si ma variable est vide, j'écris dans le terminal "C'est viiiide", sinon j'affiche sa valeur
if [ -z ${ma_variable} ]; then echo "C'est viiiide"; else echo ${ma_variable}; fi
```

#### Tester l'opposé d'une condition

Et si vous souhaitez vérifier que le fichier n'existe pas mais ne rien faire s'il existe ?

Vous ne pouvez pas écrire votre `if` en laissant votre `then` vide. Il faut donc inverser la condition.

Exemple à la place de : 
- __*If*__ mon fichier existe __*then*__ __*else*__ j'écris "n'existe pas" __*fi*__
- __*If*__ mon fichier **n'** existe **pas** __*then*__ j'écris "n'existe pas" __*fi*__

Pour inverser la condition il suffit de mettre `!` devant la condition testée.

Pour l'exemple, version code, cela donne :

```bash
# Tester si mon fichier n'existe pas, alors j'écris "n'existe pas"
if [ ! -f mon_fichier.txt ]; then echo "N'existe pas"; fi
```

## Les conseils de Polymalin 

### Itérer sur une liste de valeurs : _for_ &#x1F451; &#x1F451; &#x1F451;

La boucle `for` c'est quand vous voulez tout parcourir sans tout réécrire.

Vous prenez une suite de valeur et pour chacune vous réalisez une action.

- En français : Pour chaque valeur dans ma liste, faire une action.

Exemple : Pour chaque jour de la semaine, à faire : je me lève et je te bouscule.

- En anglais : For chaque valeur in ma liste, do une action.

Exemple : For chaque jour in la semaine, to do : je me lève et je te bouscule.

- En code : `for` valeur `in` liste`; do` action`; done`

Exemple : `for` jour `in` la semaine`; do` je dis quel jour on est `;` je me lève `;` je te bouscule `; done`

Les actions qui seront réalisées en conséquence seront :
- Lundi je me lève et je te bouscule
- Mardi je me lève et je te bouscule
- Mercredi je me lève et je te bouscule
- Jeudi je me lève et je te bouscule
- Vendredi je me lève et je te bouscule
- Samedi je me lève et je te bouscule
- Dimanche je me lève et je te bouscule

Comme pour le `if`, vous pouvez mettre n'importe quelles actions entre le `do` et le `done`.

Concernant la liste qui défini les valeurs, elle peut aussi être diversifiée, voici quelques exemples utiles :

- Sur une liste écrite à la volée

Par exemple :
```bash
# Afficher chaque nombre de la liste
for nombre in 1 2 3 4 5; do echo ${nombre}; done
# Pour chaque jour de la liste, écrire que le jour est un jour ouvré
for jour in lundi mardi mercredi jeudi vendredi; do echo ${jour}" est un jour ouvré"; done
```

- Sur le contenu d'un dossier

Par exemple :
```bash
# Pour chaque fichier de mon dossier, j'affiche les 10 premières lignes
for fichier in dossier; do head ${fichier}; done
# Pour chaque fichier texte de mon dossier courant, je compte le nombre de lignes
for texte in *.txt; do wc -l ${texte}; done
```

- Sur le résultat d'une commande

Par exemple :
```bash
# Pour chaque ligne de mon fichier, je compte le nombre de caractères
for ligne in $(cat fichier.txt); do echo ${ligne} | wc -c; done
# Pour chaque élément de ma liste (issue d'une colonne triée et rendue unique de mon_fichier.txt), je le recherche dans mon_fichier2.txt
for element in $(cut -f2 mon_fichier.txt | sort | uniq); do grep ${element} mon_fichier2.txt; done
```
