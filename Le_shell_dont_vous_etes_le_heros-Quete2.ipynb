## Les conseils de Polymalin 

### Consulter le début et la fin d'un fichier : _head_ et _tail_ &#x1F451;

Afin de ne pas afficher l'entièreté d'un fichier qui serait trop volumineux, vous pouvez utiliser les commandes `head` et `tail` qui affichent respectivement les 10 premières ou les 10 dernières lignes de votre fichier.

```bash
# Afficher les 10 premières lignes
head mon_fichier.txt
# Afficher les 10 dernières lignes
tail mon_fichier.txt
```

Si vous souhaitez afficher un autre nombre de lignes, vous pouvez utiliser l'option `-n` suivie du nombre de lignes souhaité.

```bash
# Afficher les 2 premières lignes du fichier
head -n 2 mon_fichier.txt
# Afficher les 30 dernières lignes du fichier
tail -n 30 mon_fichier.txt
```

### Rechercher dans un fichier : _grep_ &#x1F451;

Lorsque vous recherchez un mot dans un fichier, il est possible d'utiliser l'équivalent du `ctrl + f` de Linux : `grep`.

Pour cela, il suffit de lui indiquer ce que vous cherchez entre guillemets puis où le rechercher.

```bash
# Rechercher "toto" dans votre fichier
grep "toto" mon_fichier.txt
```

### Rechercher dans un fichier : _grep_ &#x1F451; &#x1F451;

`grep`, comme beaucoup de commandes Linux est dit "sensible à la casse". Ainsi, "a" et "A" sont deux caractères bien distincts pour lui. L'option `-i` permettra de le rendre insensible (quel sans coeur...). Cela ne fonctionnera malheureusement pas pour les accents. Mieux vaut donc parler en anglais dans vos fichiers.

```bash
# Rechercher id, ID, Id et iD dans mon fichier
grep -i "id" mon_fichier.txt
```

## Les conseils de Polymalin 

### Enchaîner des commandes : _pipe_ &#x1F451; &#x1F451; &#x1F451;

Lorsqu'une commande est exécutée dans le terminal, vous lui indiquez une entrée (un fichier, une valeur, parfois même rien, tout dépend de la commande). Elle vous affiche en échange un résultat qui correspond soit à la sortie standard (quand tout va bien), soit à la sortie erreur (quand tout ne va pas si bien).

Une sortie peut être exploitée de différentes façons :
- Vous n'avez pas besoin de la conserver: elle s'affiche simplement à l'écran
- Vous souhaitez la conserver: vous l'écrivez dans un fichier avec le symbole `>` vu précédemment
- Vous voulez l'utiliser pour une autre commande: vous la redirigez avec le symbole `|` (Appelé _pipe_)

Grâce au _pipe_, vous pouvez dorénavant enchainer les commandes.

Un exemple vaut mille mots :
```bash
# Je souhaite récupérer les 10 premières lignes d'un fichier
head mon_fichier.txt > fichier_intermediaire.txt
# Je veux maintenant rechercher le mot "toto" dans ces 10 lignes
grep "toto" fichier_intermediaire.txt
# Afin de ne pas écrire trop de fichiers intermédiaires encombrants, j'enchaîne mes commandes
head mon_fichier.txt | grep "toto"
# Le résultat sera le même, vous aurez juste économisé un fichier. Quand les commandes s'enchaînent et sont nombreuses, c'est un vrai gain de temps et de place !
```

### Compter des lignes : _Word Count_ &#x1F451; 

La fonction _Word Count_ est une des préférées des bioinformagiciens du royaume. Sa fonction première est de compter le nombre de mots d'un fichier. Cependant, aucun bioinformagicien n'utilise cette fonction ainsi. Ils emploient l'option `-l` permettant de compter les lignes.

```bash
# Compter le nombre de ligne de mon fichier
wc -l mon_fichier.txt
```

## Les conseils de Polymalin 

### Isoler une colonne : _cut_ &#x1F451;

Tout texte peut être coupé par colonne. C'est le rôle de la commande `cut`.

Cette commande est toujours suivie d'une option selon si vous souhaitez isoler un nombre exact de bytes `-b`, de caractères `-c` ou de champs (fields en anglais) `-f`. Ce dernier correspond à ce que vous pourriez qualifier de mot.

Ces options sont suivies d'un nombre, correspondant à ce que vous souhaitez isoler :
- `N` : Isoler le Nième byte/caractère/champs
- `N-` : Isoler les champs, à partir de N et jusqu'à la fin de la ligne
- `N-M` : Isoler les champs de N **à** M
- `-M` : Isoler les champs du premier jusqu'à M
- `N,M` : Isoler les champs N **et** M

```bash
# Isoler le champs 6
cut -f6 mon_fichier.txt
# Isoler les bytes de 4 à 9
cut -b 4-9 mon_fichier.txt
# Isoler ma ligne à partir du caractère 35
cut -c 35- mon_fichier.txt
```

### L'éditeur de flux : _sed_ &#x1F451; &#x1F451; &#x1F451;

`sed`, alias **s**tream **ed**itor, est sans doute la formule la plus puissante du bioinformagicien. Mais qui dit puissante dit aussi complexe.

A l'origine, il s'agit d'un langage à part entière. Par conséquent, vous devrez toujours avoir une commande sous le format suivant : 
```bash 
sed ''
```

Entre ces guillemets simples (simple quote), vous décrirez l'action à réaliser. Elle peut être de trois types :
- `s` : substitue une chaîne de caractères par une autre
- `d` : supprime une ligne
- `y` : substitue plusieurs caractères de façon indépendante par un ou plusieurs autres

Chacune de ces actions peut être conditionnée ou appliquée à un numéro de ligne en particulier, ainsi que sur l'itération choisie.

Quelques exemples :
```bash
# Substituer "toto" par "tata" dans tout le fichier. La substitution se fait au format :
# s/mot_a_substituer/mot_de_remplacement/l_iteration_que_vous_voulez_substituer.
# Le g indique que vous souhaitez toutes les substituer.
sed 's/toto/tata/g' mon_fichier.txt
# Supprimer une ligne contenant un caractère ou une chaîne de caractères, ici "titi"
# L'utilisation des // permet de délimiter l'expression recherchée
sed '/titi/d' mon_fichier.txt
# Modifier plusieurs caractères sur une ligne unique, sans toucher les autres
# Ici on change les "à" par "a" et les "é" par "e" sur la ligne 4
sed '4 y/àé/ae/' mon_fichier.txt
```

Il est recommandé d'afficher ce que vous modifiez à l'écran. Ainsi vous pouvez vérifier que la modification se fait correctement. A ce stade elle n'est pas sauvegardée. Ensuite, vous pouvez réécrire directement cette nouvelle version dans votre fichier avec l'option `-i`. Il s'auto-écrasera.

```bash
# Vous affichez à l'écran la substitution du 5ème "titi" de la ligne par "toto" si celle-ci contient "tata".
sed '/tata/ s/titi/toto/5' mon_fichier.txt
# Même chose mais en redirigeant le résultat dans un fichier "mon_fichier.substitue.txt" 
sed '/tata/ s/titi/toto/5' mon_fichier.txt > mon_fichier.substitue.txt
# Même chose mais en corrigeant directement le fichier
sed -i '/tata/ s/titi/toto/5' mon_fichier.txt
```

Vous avez dorénavant les bases du `sed`. Si vous souhaitez plus d'exemples et d'informations, vous pouvez consulter ce [guide encore plus complet](https://www.it-connect.fr/guide-de-survie-pour-lutilisation-de-la-commande-sed/) dans votre boule de cristal.
