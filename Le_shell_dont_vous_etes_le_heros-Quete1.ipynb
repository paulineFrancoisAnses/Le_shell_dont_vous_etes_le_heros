### Se déplacer : _Change directory_ &#x1F451;

Afin de se déplacer, vous devez utiliser la commande `cd` (_change directory_).

Suivie d'un espace puis du chemin où vous souhaitez vous rendre. Chaque sous-dossier est séparé par un "/".

```bash

# Chemin relatif - Depuis là où vous vous trouvez
cd chemin/partiel/où/vous/voulez/aller
```

## Les conseils de Polymalin 

### Copier-coller un fichier : _Copy_ &#x1F451;

Copier-coller sous Windows vous semblait facile ? Hé bien c'est encore plus simple dans un Shell !

Il vous suffit d'utiliser la commande `cp` (_copy_). Ensuite, un espace, le fichier ou dossier que vous souhaitez copier, espace, le chemin où vous souhaitez le recopier.

```bash
# Copier un fichier dans un dossier
cp mon_fichier.txt /chemin/de/destination
# Copier un fichier distant dans un dossier
cp chemin/vers/mon_fichier.txt /chemin/de/destination
# Copier un dossier, le -r (récursive) stipule que tout le contenu du dossier vient avec lui
cp -r mon_dossier /chemin/de/destination
```

### Couper-coller un fichier : _Move_ &#x1F451;

C'est comme un copier-coller mais en fait c'est un couper-coller.

Avec la commande `mv` (_move_), la structure sera la même.

```bash
# Déplacer un fichier dans un dossier
mv mon_fichier.txt /chemin/de/destination
# Déplacer un fichier distant dans un dossier
mv chemin/vers/mon_fichier.txt /chemin/de/destination
# Déplacer un dossier dans un autre
mv mon_dossier /chemin/de/destination
```

 __Le petit tips de Polymalin :__

`cp` et `mv` n'ont pas de temps à perdre. Ils ne demanderont pas si vous êtes sûr de votre choix. Ainsi, __si le fichier existe déjà, il sera écrasé__.

## Les conseils de Polymalin 

### Couper-coller un fichier : _Move_ &#x1F451; &#x1F451;

Si vous souhaitez renommer un fichier, ou un dossier, vous pouvez simplement utiliser la commande `mv`. A la place d'indiquer uniquement le chemin de destination, vous ajouterez le nouveau nom du fichier ou dossier !

```bash
# Renommer le fichier
mv mon_fichier.txt nouveau_nom.txt
# Renommer un dossier
mv mon_dossier nouveau_nom
# Déplacer et renommer le fichier
mv mon_fichier.txt dossier/de/destination/nouveau_nom.txt
# Déplacer et renommer un dossier
mv mon_dossier dossier/de/destination/nouveau_nom
```

Cette possibilité est aussi valable avec `cp` : vous pouvez créer une copie du fichier ou dossier en lui donnant un nouveau nom, ce qui permet de conserver l’original intact.

### Supprimer un fichier ou un dossier : _Remove_ &#x1F451;

&#x1F6A8; __Attention ! Encore une fois, le Shell ne vous demandera pas si vous êtes sûr avant de supprimer le fichier ou le dossier. Mal exécutée, cette commande peut vous faire perdre tous vos biens.__

Si vous souhaitez supprimer un fichier ou un dossier, vous pouvez utiliser la commande `rm` (_remove_). Ajoutez un espace puis ce que vous souhaitez supprimer. Ajoutez l'option `-r` lorsque c'est un dossier et le tour est joué !

```bash
# Supprimer un fichier
rm mon_fichier.txt
# Supprimer un dossier
rm -r mon_dossier
```

### Les formats de fichiers &#x1F451;

Il n'existe en réalité que quatre types de fichiers interprétables par l'ordinateur : Texte, compressé, exécutable et dossier.

- Le fichier texte : C'est le format de quasiment tous les fichiers. Peu importe leur extension (.txt, .csv, .fasta, .lateteatoto), le fichier est lisible et inteprétable par l'oeil humain. 

- Le fichier compressé : Destiné à prendre moins de place en mémoire, le fichier compressé devient illisible et nécessite des outils pour l'extraire en format texte.

```bash
# Extraire un fichier zip
unzip mon_fichier.zip
# Extraire un fichier gz
gunzip mon_fichier.gz
```

- Le fichier exécutable : C'est un fichier lisible par l'oeil humain mais qui recèle bien des secrets ! En effet, au même titre que les `.exe` chez Windows ou `.dmg` chez MacOS, Linux peut exécuter des fichiers `.sh` afin de réaliser l'ensemble des tâches qu'ils contiennent.

```bash
# Exécuter le fichier
./fichier_a_executer.sh
```

Un fichier exécutable peut prendre des arguments (paramètres, options, fichiers d'entrée, etc). Leur formatage est aussi diversifié que le nombre d'elfes dans ce royaume et vous avez arrêté leur recensement en 10 après J.C., c'est vous dire s'ils sont nombreux... 

Un règle à retenir cependant, il y a toujours un espace avant de commencer à écrire les arguments.
```bash
# Fichier exécuté avec l'option d'aide
./fichier_a_executer.sh -h
# Fichier exécuté avec un paramètre de chiffre 
./fichier_a_executer.sh 40
# Fichier exécuté avec une option avec un paramètre
./fichier_a_executer.sh --taille 40
```

- Le dossier : Hé oui, pour Linux un dossier est une sorte de fichier.

## Les conseils de Polymalin 

### Afficher le contenu d'un fichier : _cat_ &#x1F451;

Parfois il est nécessaire de voir le contenu d'un fichier pour en comprendre le sens. Pour cela il suffit d'utiliser la commande `cat` suivi d'un espace et du fichier à consulter.

```bash
# Afficher mon fichier
cat mon_fichier.txt
# Afficher mes deux fichiers
cat mon_fichier1.txt mon_fichier2.txt
```

### Fusionner plusieurs fichiers : _cat_ &#x1F451; &#x1F451;

Si vous affichez le contenu de plusieurs fichiers à la suite, vous pouvez ensuite le stocker dans un seul fichier. Pour cela, il suffit de rediriger la sortie avec le symbole `>`. Le contenu ne s'affichera plus à l'écran, mais sera directement écrit dans le fichier de concaténation.

```bash
# Fusionner fichier1.txt et fichier2.txt dans un fichier_fusion.txt
cat fichier1.txt fichier2.txt > fichier_fusion.txt
```
